```python
from pathlib import Path
import os, sys, subprocess, json
ROOT=Path.cwd()
if ROOT.name == 'cuadernos': ROOT=ROOT.parent
assert (ROOT/'entrenamiento/entrenar.py').exists(), 'Ubicar el proyecto completo'
os.chdir(ROOT)
```

```python
subprocess.run([sys.executable,'herramientas/descargar_dataset.py'],check=True)
```

```python
from PIL import Image
import matplotlib.pyplot as plt
fig,axes=plt.subplots(2,4,figsize=(12,6))
for row,name in enumerate(['non_fire_images','fire_images']):
    paths=sorted((ROOT/'dataset/original/fire_dataset'/name).glob('*'))
    print(name,len(paths))
    for ax,path in zip(axes[row],paths[:4]):
        with Image.open(path) as image: ax.imshow(image)
        ax.set_title(path.name); ax.axis('off')
plt.tight_layout()
```
